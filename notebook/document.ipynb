```python
## create a simple txt file
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_text = {
    "../data/text_files/python_intro.txt":"""Python is a high-level, general-purpose programming language known for its clean, readable syntax that looks similar to plain English. Created by Guido van Rossum and first released in 1991, it has become one of the most popular languages for both beginners and experienced developers across industries.

Key Features of Python

• Easy to read and write. Python uses simple English-like commands and relies on indentation (spaces) rather than curly brackets {} to organize code blocks.
• Interpreted and dynamically typed. Code runs line by line through an interpreter, and you do not need to declare variable types (like numbers or text) in advance.
• Cross-platform compatibility. Python programs run on Windows, macOS, Linux, and even small devices like the Raspberry Pi with little to no modification.
• Extensive library support. Its massive standard library and ecosystem of third-party packages let you perform complex tasks without writing everything from scratch""",

"../data/text_files/RAG_intro.txt":"""Retrieval-Augmented Generation (RAG) is an AI technique that gives a large language model (LLM) access to external documents or databases so it can answer questions with verified, up-to-date facts. Think of it as giving an AI an "open-book exam" instead of forcing it to rely strictly on what it memorized during training.

Why RAG Is Needed

Standard LLMs face three main limitations that RAG solves:
• Outdated knowledge. An LLM only knows information up to the date it was trained.
• Lack of private data. Public AI models cannot see your personal notes, company wikis, or private PDFs.
• Hallucinations. When an LLM does not know an answer, it may guess and generate confident-sounding but false information.
"""

}

for filepath, content in sample_text.items():
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(content)
print("file is created")
```

```python
###Text Loader

from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/python_intro.txt",encoding="utf-8")
document=loader.load()
print(document)
```

```python
    ###Directory Loader
from langchain_community.document_loaders import DirectoryLoader

##load all the textfiles from dir.
dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding":"utf-8"},
    show_progress=False
)

documents=dir_loader.load()
documents
```

```python


from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

##load all the textfiles from dir.
dir_loader=DirectoryLoader(
    "../data/PDF",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
   
    show_progress=False
)

pdf_documents=dir_loader.load()
pdf_documents
```
